```python
from cosmology import CosmologyParameters, CPL

pars = CosmologyParameters(
    H0=68.0,
    Omega_m=0.32,
    w0=-0.72,
    wa=-1.28,
    rd=147.0,
)

cosmo = CPL(pars)

z = 0.5

print("E(z)        =", cosmo.background.E(z))
print("H(z)        =", cosmo.background.H(z))
print("dE/dz       =", cosmo.background.dEdz(z))
print("q(z)        =", cosmo.background.q(z))
print("Dc(z) [Mpc] =", cosmo.distance.Dc(z))
print("Dl(z) [Mpc] =", cosmo.distance.Dl(z))
print("mu(z)       =", cosmo.distance.mu(z))
print("DV(z) [Mpc] =", cosmo.distance.DV(z))
print("rd [Mpc]    =", cosmo.sound_horizon.rd)
```
